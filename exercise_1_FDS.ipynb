```python
from cryptography.fernet import Fernet

# Generate key and encrypt data
key = Fernet.generate_key()
cipher_suite = Fernet(key)

plain_text = b"Rajalakshmi Engineering College"

# Encrypt data
cipher_text = cipher_suite.encrypt(plain_text)

# Decrypt data
decrypted_text = cipher_suite.decrypt(cipher_text)

# Display the results
print("Original Data:", plain_text)
print("Encrypted Data:", cipher_text)
print("Decrypted Data:", decrypted_text)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create the dataset
data = {
    'Year': [2015, 2016, 2017, 2018, 2019,
             2020, 2021, 2022, 2023, 2024],
    'Job_Postings': [1200, 1800, 2600, 3900, 5500,
                     7200, 9800, 11500, 13200, 15000]
}

# Create a DataFrame
df = pd.DataFrame(data)

# Display the dataset
print("Data Science Job Postings:")
print(df)

# Plot the trend
plt.figure(figsize=(10, 5))
plt.plot(df['Year'], df['Job_Postings'],
         marker='o', linestyle='-')

plt.title("Trend of Data Science Job Postings (2015-2024)")
plt.xlabel("Year")
plt.ylabel("Number of Job Postings")
plt.xticks(df['Year'])
plt.grid(True)
plt.show()

# Analyze the trend
increase = df['Job_Postings'].iloc[-1] - df['Job_Postings'].iloc[0]
percentage = (increase / df['Job_Postings'].iloc[0]) * 100

print("Job postings in 2015:", df['Job_Postings'].iloc[0])
print("Job postings in 2024:", df['Job_Postings'].iloc[-1])
print("Total increase:", increase)
print("Percentage increase:", percentage, "%")
```

```python
import matplotlib.pyplot as plt

# Data
roles = ['Data Analyst', 'Data Engineer', 'Data Scientist',
         'ML Engineer', 'Business Analyst']
counts = [300, 500, 450, 200, 150]

# Create bar chart
plt.figure(figsize=(10, 5))
plt.bar(roles, counts)

# Add title and labels
plt.title('Distribution of Data Science Roles')
plt.xlabel('Role')
plt.ylabel('Count')

plt.show()

# Display role counts
for role, count in zip(roles, counts):
    print(role, ":", count)
```

```python
import pandas as pd

# Structured Data Example
structured_data = pd.DataFrame({
    'Product_ID': [101, 102, 103],
    'Product_Name': ['Laptop', 'Mobile', 'Tablet'],
    'Price': [50000, 20000, 15000]
})

print("Structured Data:\n", structured_data)

# Unstructured Data Example
unstructured_data = """
Customer Review:
The laptop has excellent performance and battery life.
The mobile camera is very good.
"""

print("\nUnstructured Data:\n", unstructured_data)

# Semi-structured Data Example (JSON)
semi_structured_data = {
    'Product_ID': 101,
    'Product_Name': 'Laptop',
    'Specifications': {
        'RAM': '16GB',
        'Storage': '512GB SSD'
    }
}

print("\nSemi-structured Data:\n", semi_structured_data)
```

```python
from cryptography.fernet import Fernet

# Generate an encryption key
key = Fernet.generate_key()
cipher = Fernet(key)

# Sensitive data (Example: Bank Account Details)
sensitive_data = b"Account Number: 1234567890"

# Encrypt the sensitive data
encrypted_data = cipher.encrypt(sensitive_data)

# Decrypt the encrypted data
decrypted_data = cipher.decrypt(encrypted_data)

# Display the results
print("Original Data:", sensitive_data.decode())
print("Encrypted Data:", encrypted_data.decode())
print("Decrypted Data:", decrypted_data.decode())
```
